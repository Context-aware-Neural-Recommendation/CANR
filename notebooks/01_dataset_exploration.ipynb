```python
import pandas as pd

articles = pd.read_csv(
    "../data/raw/articles.csv",
    nrows=5
)

customers = pd.read_csv(
    "../data/raw/customers.csv",
    nrows=5
)

transactions = pd.read_csv(
    "../data/raw/transactions_train.csv",
    nrows=5
)

print("Articles:")
display(articles)

print("Customers:")
display(customers)

print("Transactions:")
display(transactions)
```

```python
print("ARTICLES COLUMNS")
print(articles.columns.tolist())

print("\nCUSTOMERS COLUMNS")
print(customers.columns.tolist())

print("\nTRANSACTIONS COLUMNS")
print(transactions.columns.tolist())
```

```python
print("ARTICLES")
print(articles.dtypes)

print("\nCUSTOMERS")
print(customers.dtypes)

print("\nTRANSACTIONS")
print(transactions.dtypes)
```

```python
print("Articles missing values:")
print(articles.isnull().sum())

print("\nCustomers missing values:")
print(customers.isnull().sum())

print("\nTransactions missing values:")
print(transactions.isnull().sum())
```
